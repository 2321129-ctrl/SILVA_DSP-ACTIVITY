# DSP Activity 1: Plotting of Sequences
Plots an arbitrary finite-duration sequence and the standard elementary sequences (impulse, step, ramp, sine, cosine) using `numpy` and `matplotlib` (based on Activity #2).

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Arbitrary finite-duration sequence

```python
n = np.arange(-5, 6)
x = np.array([0, 5, 6, 7, 9, 0, 0, -5, -9, -7, -6])
print('n:', n)
print('x:', x)

plt.stem(n, x)
plt.title('Arbitrary finite-duration sequence')
plt.xlabel('sampling interval n')
plt.ylabel('sample x(n)')
plt.grid(True)
plt.show()
```

```python
n = np.arange(-4, 5)
xn = (n == 0).astype(int)
plt.stem(n, xn)
plt.title('Unit impulse')
plt.xlabel('Sampling Interval n')
plt.ylabel('Sample')
plt.grid(True)
plt.show()
```

```python
n = np.arange(-4, 5)
xn = (n >= 0).astype(int)
plt.stem(n, xn)
plt.title('Unit step')
plt.xlabel('Sampling Interval n')
plt.ylabel('Sample')
plt.grid(True)
plt.show()
```

```python
n = np.arange(-4, 5)
xn = n * (n >= 0)
plt.stem(n, xn)
plt.title('Unit ramp')
plt.xlabel('Sampling Interval n')
plt.ylabel('Sample')
plt.grid(True)
plt.show()
```

```python
n = np.arange(-6.2, 6.3, 0.1)

fig, axs = plt.subplots(1, 2, figsize=(14, 4))
axs[0].stem(n, np.sin(n))
axs[0].set_title('Sine wave  sin(n)')
axs[1].stem(n, np.cos(n))
axs[1].set_title('Cosine wave  cos(n)')
for a in axs:
    a.set_xlabel('Sampling Interval n')
    a.set_ylabel('Sample')
    a.grid(True)
plt.tight_layout()
plt.show()
```
